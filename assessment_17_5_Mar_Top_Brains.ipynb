```python
#Scenario 1: Online Course Completion Prediction
#We want to classify whether a student will complete an online course (1) or drop out (0) based on two features:
#Videos Watched – number of course videos the student watched
#Time Spent on Platform – total minutes spent on the learning platform

import numpy as np

# Training data
X = np.array([
    [2,15],
    [3,20],
    [8,60],
    [9,75]
])
y = np.array([[0],[0],[1],[1]])

# Normalize data
X = X / np.max(X, axis=0)

# Initialize weights
np.random.seed(1)
w1 = np.random.rand(2,3)
w2 = np.random.rand(3,1)

# Sigmoid function
def sigmoid(x):
    return 1/(1+np.exp(-x))

# Derivative of sigmoid
def sigmoid_derivative(x):
    return x*(1-x)

# Training
for i in range(5000):
    # Forward propagation
    hidden_input = np.dot(X,w1)
    hidden_output = sigmoid(hidden_input)
    final_input = np.dot(hidden_output,w2)
    output = sigmoid(final_input)

    # Error
    error = y - output

    # Backpropagation
    d_output = error * sigmoid_derivative(output)
    error_hidden = d_output.dot(w2.T)
    d_hidden = error_hidden * sigmoid_derivative(hidden_output)

    # Update weights
    w2 += hidden_output.T.dot(d_output)
    w1 += X.T.dot(d_hidden)

print("Predictions:")
print(output)
```
