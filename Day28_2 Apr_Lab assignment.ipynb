```python
#📘 LAB 1: Python Fundamentals + File Handling
#🎯 Objective:
#Build a Student Result Processing System

#🧩 Tasks:
#Take input for 5 students (name + marks)

#Calculate percentage & grade

#Store results in a .txt file

#Read file and display toppers

📤 Expected Output:
#Structured file output

#Sorted results





def calculate_grade(percentage):
    if percentage >= 90:
        return 'A'
    elif percentage >= 75:
        return 'B'
    elif percentage >= 50:
        return 'C'
    else:
        return 'Fail'

students = []

# Input
for i in range(5):
    name = input("Enter name: ")
    marks = float(input("Enter marks: "))
    percentage = marks
    grade = calculate_grade(percentage)
    students.append((name, percentage, grade))

# Sort by marks
students.sort(key=lambda x: x[1], reverse=True)

# Write to file
with open("results.txt", "w") as f:
    for s in students:
        f.write(f"{s[0]} - {s[1]}% - {s[2]}\n")

# Read and display topper
print("\nTopper:")
print(students[0])
```

```python
# 📘 LAB 2: Data Types + Control Flow

# 🎯 Objective:
# Build a Data Analyzer Tool

# 🧩 Tasks:
# Input a mixed list

# Separate integers, floats, strings

# Count frequency of each type

# Print summary using loops




data = [10, "hello", 5.5, 20, "world", 3.14, 10]

ints, floats, strings = [], [], []

for item in data:
    if isinstance(item, int):
        ints.append(item)
    elif isinstance(item, float):
        floats.append(item)
    else:
        strings.append(item)

print("Integers:", ints)
print("Floats:", floats)
print("Strings:", strings)

print("\nCounts:")
print("Integers:", len(ints))
print("Floats:", len(floats))
print("Strings:", len(strings))
```

```python
# 📘 LAB 3: Functions + Recursion

# 🎯 Objective:
# Create a Math Utility Module

# 🧩 Tasks:
# Factorial using recursion

# Fibonacci using recursion

# Lambda for square & cube

# Compare execution time



import time

def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)

def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

square = lambda x: x*x
cube = lambda x: x*x*x

# Time comparison
start = time.time()
print("Factorial:", factorial(5))
print("Fibonacci:", fibonacci(6))
end = time.time()

print("Execution Time:", end-start)

print("Square:", square(4))
print("Cube:", cube(3))
```

```python
# 📘 LAB 4: Regex + File Processing

# 🎯 Objective:
# Build an Email Extractor Tool

# 🧩 Tasks:
# Read a text file

# Extract valid emails using regex

# Count unique emails

# Save results in CSV


# 📘 LAB 4: Email Extractor Tool (FINAL VERSION)

import re
import csv
import os

FILE_NAME = "data.txt"

# Step 1: Check if file exists, otherwise create it
if not os.path.exists(FILE_NAME):
    print("data.txt not found. Creating sample file...\n")

    with open(FILE_NAME, "w") as f:
        f.write("""hello@gmail.com
test123@yahoo.com
invalid-email
user.name@company.in
another_user@domain.org
hello@gmail.com
""")

# Step 2: Read file
with open(FILE_NAME, "r") as f:
    text = f.read()

# Step 3: Extract emails using regex
pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
emails = re.findall(pattern, text)

# Step 4: Get unique emails
unique_emails = list(set(emails))

# Step 5: Display results
print("📧 All Extracted Emails:")
for email in emails:
    print(email)

print("\n📊 Unique Emails:")
for email in unique_emails:
    print(email)

print(f"\nTotal Emails Found: {len(emails)}")
print(f"Unique Emails Count: {len(unique_emails)}")

# Step 6: Save to CSV
with open("emails.csv", "w", newline='') as f:
    writer = csv.writer(f)
    writer.writerow(["Emails"])

    for e in unique_emails:
        writer.writerow([e])

print("\n✅ Emails saved to 'emails.csv'")
```

```python
# 📘 LAB 5: Exception Handling

# 🎯 Objective:
# Develop a Robust Calculator

# 🧩 Tasks:
# Handle invalid input

# Handle division by zero

# Use try–except–finally

# Create custom exception


class NegativeNumberError(Exception):
    pass

try:
    a = float(input("Enter number 1: "))
    b = float(input("Enter number 2: "))

    if b == 0:
        raise ZeroDivisionError

    if a < 0 or b < 0:
        raise NegativeNumberError("Negative numbers not allowed")

    print("Division:", a / b)

except ZeroDivisionError:
    print("Error: Division by zero")

except NegativeNumberError as e:
    print(e)

except ValueError:
    print("Invalid input")

finally:
    print("Calculation complete")
```

```python
# 📘 LAB 6: OOP + Inheritance

# 🎯 Objective:
# Design an Employee Management System

# 🧩 Tasks:
# Base class: Employee

# Derived: Manager, Developer

# Override salary method

# Display employee details



class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def get_salary(self):
        return self.salary

class Manager(Employee):
    def get_salary(self):
        return self.salary + 5000

class Developer(Employee):
    def get_salary(self):
        return self.salary + 3000

m = Manager("Arnav", 50000)
d = Developer("Archana", 40000)

print(m.name, m.get_salary())
print(d.name, d.get_salary())
```

```python
# 📘 LAB 7: Decorators + Generators

# 🎯 Objective:
# Optimize performance using advanced Python

# 🧩 Tasks:
# Create decorator for logging

# Create generator for large data

# Compare memory usage


import sys

def logger(func):
    def wrapper(*args):
        print("Function called:", func.__name__)
        return func(*args)
    return wrapper

@logger
def display():
    print("Hello World")

display()

# Generator
def generate_numbers(n):
    for i in range(n):
        yield i

gen = generate_numbers(1000000)
print("Memory efficient generator created")
```

```python
# 📘 LAB 8: Multithreading

# 🎯 Objective:
# Simulate Parallel File Downloader

# 🧩 Tasks:
# Download multiple files using threads

# Measure execution time

# Compare with sequential execution


import threading
import time

def download(file):
    print(f"Downloading {file}")
    time.sleep(2)
    print(f"Downloaded {file}")

files = ["file1", "file2", "file3"]

threads = []

start = time.time()

for f in files:
    t = threading.Thread(target=download, args=(f,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

end = time.time()

print("Time:", end-start)
```

```python
# 📘 LAB 9: NumPy + Pandas

# 🎯 Objective:
# Perform Data Analysis

# 🧩 Tasks:
# Load dataset

# Handle missing values

# Perform filtering

# Compute mean, median, aggregation



import pandas as pd

df = pd.read_csv("data.csv")

# Handle missing values
df.fillna(df.mean(numeric_only=True), inplace=True)

# Filtering
filtered = df[df['marks'] > 50]

print("Filtered Data:\n", filtered)

print("Mean:", df['marks'].mean())
print("Median:", df['marks'].median())

print("Aggregation:\n", df.groupby('name').mean())
```

```python
# 📘 LAB 10: Mini Project (FINAL)

# 🎯 Objective:
# Build a Student Management System (End-to-End)

# 🧩 Features:
# Add student

# Search student

# Store in CSV

# Display analytics (topper, avg marks)



# LAB 4 - EMAIL EXTRACTOR TOOL (ADVANCED VERSION)

import re
import csv
import os

FILE_NAME = "data.txt"
CSV_NAME = "emails.csv"


# Step 1: Ensure file exists (auto-create if missing)
def create_sample_file():
    if not os.path.exists(FILE_NAME):
        print(" data.txt not found. Creating sample file...\n")
        with open(FILE_NAME, "w") as f:
            f.write("""hello@gmail.com
test123@yahoo.com
invalid-email
user.name@company.in
another_user@domain.org
hello@gmail.com
""")
        print(" Sample file created!\n")


# Step 2: Read file
def read_file():
    with open(FILE_NAME, "r") as f:
        return f.read()


# Step 3: Extract emails using regex
def extract_emails(text):
    pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
    return re.findall(pattern, text)


# Step 4: Get unique emails
def get_unique_emails(email_list):
    return list(set(email_list))


# Step 5: Save emails to CSV
def save_to_csv(emails):
    with open(CSV_NAME, "w", newline='') as f:
        writer = csv.writer(f)
        writer.writerow(["Email"])

        for email in emails:
            writer.writerow([email])


# Step 6: Display results
def display_results(all_emails, unique_emails):
    print(" All Extracted Emails:")
    for e in all_emails:
        print(e)

    print("\n Unique Emails:")
    for e in unique_emails:
        print(e)

    print(f"\n Total Emails Found: {len(all_emails)}")
    print(f" Unique Emails Count: {len(unique_emails)}")


# MAIN FUNCTION
def main():
    print("===== Email Extractor Tool =====\n")

    create_sample_file()

    text = read_file()

    all_emails = extract_emails(text)

    unique_emails = get_unique_emails(all_emails)

    save_to_csv(unique_emails)

    display_results(all_emails, unique_emails)

    print(f"\n Emails saved successfully in '{CSV_NAME}' file")


# Run program
if __name__ == "__main__":
    main()
```
